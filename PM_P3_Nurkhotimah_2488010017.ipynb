# Latihan 1

```python
import pandas as pd, seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
print(df.shape)
df.head()
df.info()

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

# Latihan 2

```python
df[['total_bill']].hist(figsize=(9,6), bins=15)
plt.show()
sns.boxplot(data=df[['tip']])
plt.title('Deteksi Outlier');
plt.show()
```

# Latihan 3

```python
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')
plt.title('Scatter Plot Total Bill vs Tip')
plt.show()
```

# Latihan 4

```python
num = df.select_dtypes(include='number')
sns.heatmap(num.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

1.   EDA penting karena membantu kita memahami kondisi dan karakteristik data sebelum digunakan untuk membangun model machine learning. Dengan EDA, kita dapat mengetahui pola atau hubungan antarvariabel, melihat adanya data kosong, outlier, serta memahami distribusi data. Hasil EDA juga dapat membantu menentukan fitur yang paling relevan dan metode preprocessing yang diperlukan sehingga model yang dibuat dapat bekerja dengan lebih baik.
2.
*   Histogram digunakan untuk melihat distribusi atau sebaran suatu variabel, misalnya apakah data banyak terkumpul pada nilai tertentu atau cenderung miring ke kanan/kiri.
*   Box plot digunakan untuk melihat median, rentang data, penyebaran, serta outlier. Jadi, box plot lebih mudah digunakan untuk mendeteksi nilai yang terlalu jauh dari data lainnya.
*   Scatter plot digunakan untuk melihat hubungan atau pola antara dua variabel numerik. Dari scatter plot kita dapat mengetahui apakah hubungan kedua variabel cenderung positif, negatif, atau tidak memiliki pola yang jelas.
